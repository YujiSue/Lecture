```
#@markdown #### <font color="darkred">実行ボタンを押して</font>教師データを作成しよう
import sys
if 'initialized' not in locals():
  print("初期設定が行われていないか、設定後に再度「ドライブにコピー」を実行した可能性があります")
  sys.exit()

import base64
import glob
import numpy as np
from IPython.display import HTML, clear_output
from keras.utils import load_img, img_to_array, to_categorical

# 学習用の教師データ作成
学習用画像 = []
診断結果 = []
画像サイズ = (300, 300, 1)

# 健常画像
画像ファイルリスト = glob.glob('/content/Normal/*.png')
for 画像ファイル in 画像ファイルリスト:
  print(f'{画像ファイル}読み込み中')
  画像 = load_img(画像ファイル, color_mode = "grayscale", target_size=画像サイズ[:2])
  学習用画像.append(img_to_array(画像))
  診断結果.append(0) # 健常 = 0

# 心拡大画像
画像ファイルリスト = glob.glob('/content/Cardio/*.png')
for 画像ファイル in 画像ファイルリスト:
  print(f'{画像ファイル}読み込み中')
  画像 = load_img(画像ファイル, color_mode = "grayscale", target_size=画像サイズ[:2])
  学習用画像.append(img_to_array(画像))
  診断結果.append(1) # 心拡大 = 1

# 大規模計算用に形式変換
学習用画像 = np.array(学習用画像)
学習用画像 = 学習用画像.astype('float32')
学習用画像 = 学習用画像 / 255.0
診断結果 = np.array(診断結果)
診断結果 = to_categorical(診断結果, 2)
clear_output()

#############################################################
with open("test.png", "rb") as image_file:
  encoded = base64.b64encode(image_file.read()).decode()
display(HTML(f'''
<h2>取り込んだ学習用画像</h2>
<table>
  <tr><th>画像</th><th>枚数</th></tr>
  <tr><td>健常画像</td><td>{int(np.sum(診断結果[:, 0]))}</td></tr>
  <tr><td>心拡大画像</td><td>{int(np.sum(診断結果[:, 1]))}</td></tr>
  <tr><td>計</td><td>{len(学習用画像)}</td></tr>
  </table>
<div>
<h2>取り分けたテスト用画像</h2>
<img src="data:image/jpeg;base64,{encoded}" width="300" alt=""/>
</div>
'''))
```

```
#@title  <font color="darkred">実行して</font>みよう
import sys
if 'initialized' not in locals():
  print("初期設定が行われていないか、設定後に再度「ドライブにコピー」を実行した可能性があります")
  sys.exit()

from ysaidoc import ニューラルネット

サンプルAI = ニューラルネット()

# 元の画像を300 x 300マスの情報とする
# AIに一度に１枚ずつ渡すので入力情報を (300, 300, 1)と表記

# 畳み込み層 (１層目)
サンプルAI.畳み込み層の追加(32, 12, (4,4), "relu", 入力画像サイズ = (300,300,1))
# プーリング層 (１層目)
サンプルAI.プーリング層の追加((3,3), (2,2))

# 畳み込み層 (２層目)：
サンプルAI.畳み込み層の追加(64, 5, (1,1), "relu")
# プーリング層 (２層目)
サンプルAI.プーリング層の追加((3,3), (2,2))

# 畳み込み層 (３層目)
サンプルAI.畳み込み層の追加(128, 3, (1,1), "relu")
# 畳み込み層 (４層目)
サンプルAI.畳み込み層の追加(128, 3, (1,1), "relu")
# 畳み込み層 (５層目)
サンプルAI.畳み込み層の追加(64, 3, (1,1), "relu")
# プーリング層 (３層目)
サンプルAI.プーリング層の追加((3,3), (2,2))

# 情報の統合
サンプルAI.情報の統合()
# 全結合層（１層目）
サンプルAI.全結合層の追加(4096, drop=0.5)
# 全結合層（２層目）
サンプルAI.全結合層の追加(4096, drop=0.5)

# 出力層（最終ノード数は２）
サンプルAI.全結合層の追加(2, 'softmax') # <- softmaxはReLUとは異なる活性化関数の１種

## 上記の構造でニューラルネットを構築
サンプルAI.構築()

## 確認
サンプルAI.構造を表示()
```

### 最初の畳み込み層で、300 x 300マスの１枚画に対して、32通りのカーネル（それぞれ異なる特徴を抽出可能）を使って 75 x 75 マスに畳み込んだ小画像を 32 セット作成する
(300, 300, 1) => (75, 75, 32)

### プーリング層でさらに小さな画像に集約（枚数は不変）
(75, 75, 32) => (37, 37, 32)

### 次の畳み込み層で、さらにカーネル数を増やす
(37, 37, 32) => (37, 37, 64)

...

### 全結合層ではその時点での総ノード数をそのまま使用
8 * 8 * 64 = 4096

### 出力層(dense_11)では、疾患の有りか無しかの２択の答えが出ればよいので、ノード数２に集約
